```python
def cap_outliers_iqr(df, columns):
    for col in columns:
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)

        IQR = Q3 - Q1

        lower_limit = Q1 - 1.5 * IQR
        upper_limit = Q3 + 1.5 * IQR

        df[col] = df[col].clip(lower=lower_limit, upper=upper_limit)

    return df
```

```python
columns_to_cap = [
    'average_cost_for_two',
    'votes'
]

df = cap_outliers_iqr(df, columns_to_cap)
```

```python
# Analyze the distribution of aggregate ratings and determine the most common rating range.
```

```python
bins = [0, 1, 2, 3, 4, 5]
labels = ['0–1', '1–2', '2–3', '3–4', '4–5']
df['rating_range'] = pd.cut(
    df['aggregate_rating'],
    bins=bins,
    labels=labels,
    include_lowest=True
)
rating_counts = df['rating_range'].value_counts().sort_index()
rating_percentage = (
    rating_counts / rating_counts.sum() * 100
).round(2)
rating_analysis = pd.DataFrame({
    'Restaurant Count': rating_counts,
    'Percentage': rating_percentage
})
print("Rating Distribution:")
print(rating_analysis)
most_common_range = rating_counts.idxmax()
most_common_count = rating_counts.max()
most_common_percentage = rating_percentage[most_common_range]
print("\nMost Common Rating Range:", most_common_range)
print("Number of Restaurants:", most_common_count)
print("Percentage:", most_common_percentage, "%")
plt.figure(figsize=(8, 5))
ax = sns.barplot(
    x=rating_analysis.index,
    y=rating_analysis['Restaurant Count']
)
plt.title('Distribution of Aggregate Ratings')
plt.xlabel('Rating Range')
plt.ylabel('Number of Restaurants')
for container in ax.containers:
    ax.bar_label(container)
plt.tight_layout()
plt.show()
```

## Insights:- 
- The rating range with the highest number of restaurants is the most common rating category, representing the dominant customer-rating segment in the dataset.

```python
## Identify the most common combinations of cuisines in the dataset.
```

```python
cuisine_data = df['cuisines'].dropna()
cuisine_combinations = (
    cuisine_data
    .str.split(',')
    .apply(lambda x: ', '.join(sorted([c.strip() for c in x])))
)
combination_counts = cuisine_combinations.value_counts()
top_combinations = combination_counts.head(10)
print("Top 10 Most Common Cuisine Combinations:")
print(top_combinations)

# Visualization
plt.figure(figsize=(10, 6))
ax = sns.barplot(
    x=top_combinations.values,
    y=top_combinations.index
)
plt.title('Top 10 Most Common Cuisine Combinations')
plt.xlabel('Number of Restaurants')
plt.ylabel('Cuisine Combination')
for container in ax.containers:
    ax.bar_label(container)
plt.tight_layout()
plt.show()
```

## Insignts:- 
- The analysis identifies the most frequently offered cuisine combinations, helping understand which cuisines are commonly paired together by restaurants.

```python
# Determine if certain cuisine combinationsb tend to have higher ratings.
```

```python
data = df[['cuisines', 'aggregate_rating']].dropna()
data['cuisine_combination'] = (
    data['cuisines']
    .str.split(',')
    .apply(lambda x: ', '.join(sorted([c.strip() for c in x])))
)
cuisine_rating = (
    data.groupby('cuisine_combination')
    .agg(
        restaurant_count=('cuisine_combination', 'size'),
        average_rating=('aggregate_rating', 'mean')
    )
    .reset_index()
)
# Consider only combinations with at least 10 restaurants
# This avoids misleading results from combinations with only 1–2 restaurants
cuisine_rating = cuisine_rating[
    cuisine_rating['restaurant_count'] >= 10
]
cuisine_rating = cuisine_rating.sort_values(
    'average_rating',
    ascending=False
)
top_10 = cuisine_rating.head(10)
print("Top 10 Cuisine Combinations by Average Rating:")
print(top_10.to_string(index=False))
# Visualization
plt.figure(figsize=(12, 5))
ax = sns.barplot(
    data=top_10,
    x='average_rating',
    y='cuisine_combination'
)
plt.title('Top Cuisine Combinations by Average Rating')
plt.xlabel('Average Rating')
plt.ylabel('Cuisine Combination')
for container in ax.containers:
    ax.bar_label(container, fmt='%.2f')
plt.xlim(0, 5)
plt.tight_layout()
plt.show()
```

## Insights:- 
- Certain cuisine combinations have higher average ratings than others, indicating that the combination of cuisines may influence customer satisfaction.

```python
## Analyze the text reviews to identify the most common positive and negative keywords.
```

```python
import re
from collections import Counter
text = ' '.join(df['rating_text'].dropna().astype(str)).lower()
words = re.findall(r'\b[a-z]+\b', text)
stopwords = {
    'the', 'and', 'is', 'to', 'of', 'a', 'in', 'for',
    'it', 'was', 'this', 'with', 'on', 'very', 'that',
    'my', 'we', 'i', 'are', 'but', 'not'
}
words = [word for word in words if word not in stopwords]
common_words = Counter(words).most_common(20)
print("Most Common Keywords:")
for word, count in common_words:
    print(word, ":", count)
```

## Insights:- 
- Positive reviews commonly contain words related to food quality and service, while negative reviews highlight issues such as poor service, high prices, or unsatisfactory food.
